```python
import sqlite3
import pandas as pd

# Connect to database
conn = sqlite3.connect("../database/healthcare.db")

# Query ALL 17 years of health expenditure for Argentina
argentina_query = """
SELECT 
    country,
    year,
    life_expectancy,
    health_expenditure,
    population
FROM health_metrics
WHERE country = 'Argentina' 
  AND health_expenditure IS NOT NULL
ORDER BY year ASC;
"""

argentina_df = pd.read_sql_query(argentina_query, conn)
print(f"Total entries found: {len(argentina_df)}")
print(argentina_df)

conn.close()
```
